```python
import pandas as pd

# Correct filename for the CSV data
filename = 'placement_predict_50k Dataset.csv'

df = pd.read_csv(filename)

print("Dataset shape:", df.shape)
print(df.head())
```

```python
print(df.columns.tolist())
```

```python
print(df["PlacementStatus"].value_counts())
```

```python
target = "PlacementStatus"
```

```python
X = df.drop(columns=[target])
```

```python
y = df[target]
```

```python
print("X shape:", X.shape)
```

```python
print("y shape:", y.shape)
```

```python
print("\nTarget distribution:")
```

```python
print(y.value_counts())
```

```python
X = pd.get_dummies(X, drop_first=True)
```

```python
print("Encoded X shape:", X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
```

```python
print("Missing values in X:")
print(X.isna().sum().sort_values(ascending=False).head(20))

print("\nTotal missing values:", X.isna().sum().sum())
```

```python
X = X.replace([np.inf, -np.inf], np.nan)

print("Missing values after replacing infinity:",
      X.isna().sum().sum())
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_val_imputed = imputer.transform(X_val)

print("Training missing values:",
      np.isnan(X_train_imputed).sum())

print("Validation missing values:",
      np.isnan(X_val_imputed).sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_val_scaled = scaler.transform(X_val_imputed)

print("X_train_scaled:", X_train_scaled.shape)
print("X_val_scaled:", X_val_scaled.shape)
```

```python
C_values = [0.001, 0.01, 0.1, 1, 10, 100]
```

```python
def compare_regularisation(X_train, y_train, X_val, y_val, C_values):
    pass
```

```python
results = []
```

```python
for C in C_values:

        # LASSO - L1
        lasso = LogisticRegression(
            penalty="l1",
            C=C,
            solver="liblinear",
            max_iter=5000,
            random_state=42
        )

        lasso.fit(X_train_scaled, y_train)

        lasso_pred = lasso.predict(X_val_scaled)

        lasso_acc = accuracy_score(
            y_val,
            lasso_pred
        )

        lasso_nonzero = np.sum(
            np.abs(lasso.coef_) > 1e-6
        )

        # RIDGE - L2
        ridge = LogisticRegression(
            penalty="l2",
            C=C,
            solver="liblinear",
            max_iter=5000,
            random_state=42
        )

        ridge.fit(X_train_scaled, y_train)

        ridge_pred = ridge.predict(X_val_scaled)

        ridge_acc = accuracy_score(
            y_val,
            ridge_pred
        )

        ridge_nonzero = np.sum(
            np.abs(ridge.coef_) > 1e-6
        )

        # ELASTIC NET - L1 + L2
        elastic = LogisticRegression(
            penalty="elasticnet",
            C=C,
            l1_ratio=0.5,
            solver="saga",
            max_iter=5000,
            random_state=42
        )

        elastic.fit(X_train_scaled, y_train)

        elastic_pred = elastic.predict(X_val_scaled)

        elastic_acc = accuracy_score(
            y_val,
            elastic_pred
        )

        elastic_nonzero = np.sum(
            np.abs(elastic.coef_) > 1e-6
        )

        results.append({
            "C": C,
            "Lasso_Accuracy": lasso_acc,
            "Lasso_NonZeroCoefficients": lasso_nonzero,
            "Ridge_Accuracy": ridge_acc,
            "Ridge_NonZeroCoefficients": ridge_nonzero,
            "ElasticNet_Accuracy": elastic_acc,
            "ElasticNet_NonZeroCoefficients": elastic_nonzero
        })
```

```python
results_df = pd.DataFrame(results)
print(results_df)
```

```python
best_lasso = results_df.loc[
    results_df["Lasso_Accuracy"].idxmax()
]

print("Best Lasso:")
print(best_lasso)

best_ridge = results_df.loc[
    results_df["Ridge_Accuracy"].idxmax()
]

print("Best Ridge:")
print(best_ridge)

best_elastic = results_df.loc[
    results_df["ElasticNet_Accuracy"].idxmax()
]

print("Best Elastic Net:")
print(best_elastic)
```

```python
plt.figure(figsize=(8, 5))
```

```python
plt.semilogx(
    results_df["C"],
    results_df["Lasso_Accuracy"],
    marker="o",
    label="Lasso"
)

plt.semilogx(
    results_df["C"],
    results_df["Ridge_Accuracy"],
    marker="o",
    label="Ridge"
)

plt.semilogx(
    results_df["C"],
    results_df["ElasticNet_Accuracy"],
    marker="o",
    label="Elastic Net"
)

plt.xlabel("C")
plt.ylabel("Validation Accuracy")
plt.title("Validation Accuracy vs C")
plt.legend()
plt.grid(True)
plt.show()
```
